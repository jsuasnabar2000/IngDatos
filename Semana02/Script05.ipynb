# Módulo MatplotLib
Matplotlib es una biblioteca de Python especializada en la creación de gráficos y visualizaciones de datos. Permite generar desde gráficos simples de líneas y barras hasta representaciones más complejas como histogramas, diagramas de dispersión, mapas de calor y figuras tridimensionales. Su principal ventaja es la flexibilidad: ofrece una amplia gama de funciones para personalizar colores, estilos, etiquetas y escalas, lo que la convierte en una herramienta esencial en ciencia de datos, análisis estadístico e investigación científica. Además, se integra fácilmente con otras librerías como NumPy y Pandas, facilitando la visualización directa de datos numéricos y tabulares.

## Funciones principales
Matplotlib ofrece una amplia gama de funciones para distintos tipos de gráficos:

**Gráficos de líneas:** plt.plot(x, y)

**Gráficos de dispersión:** plt.scatter(x, y)

**Gráficos de barras:** plt.bar(x, y)

**Histogramas:** plt.hist(datos, bins=10)

**Gráficos circulares (pie chart):** plt.pie(valores, labels=etiquetas)

**Imágenes y mapas de calor:** plt.imshow(matriz, cmap='hot')

**Subplots:** plt.subplots() para organizar múltiples gráficos en una misma figura.

## Personalización
**Colores y estilos:** Se pueden definir con parámetros como color, linestyle, marker.

**Leyendas:** plt.legend() para identificar series de datos.

**Configuración global:** Uso de rcParams y hojas de estilo (plt.style.use('ggplot')).

**Orientación a objetos:** Crear Figure y Axes con plt.subplots() para mayor control.

```python
# Ejemplo 01
import matplotlib.pyplot as plt

# Datos simples
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

# Crear gráfico de líneas
plt.plot(x, y)

# Personalizar
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.title("Ejemplo básico")

# Mostrar
plt.show()
```

```python
# Ejemplo 02 - Gráfico de Barras
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C', 'D']
valores = [3, 7, 5, 9]

plt.bar(categorias, valores, color='skyblue')
plt.title("Gráfico de Barras")
plt.show()
```

```python
# Ejemplo 03 - Gráfico de Dispersión
import matplotlib.pyplot as plt

x = [5, 7, 8, 7, 6, 9, 5, 6]
y = [87, 95, 98, 96, 90, 100, 88, 91]

plt.scatter(x, y, color='red')
plt.title("Gráfico de Dispersión")
plt.xlabel("Variable X")
plt.ylabel("Variable Y")
plt.show()
```

```python
# Ejemplo 04 - Histograma
import matplotlib.pyplot as plt
import numpy as np

datos = np.random.randn(1000)

plt.hist(datos, bins=30, color='green', alpha=0.7)
plt.title("Histograma")
plt.show()
```

```python
# Ejemplo 05 - Gráficos Pastel
import matplotlib.pyplot as plt

etiquetas = ['Python', 'Java', 'C++', 'JavaScript']
valores = [215, 130, 245, 210]

plt.pie(valores, labels=etiquetas, autopct='%1.1f%%', startangle=90)
plt.title("Gráfico Circular")
plt.show()
```

```python
# Ejemplo 06 - Mapa de Calor
import matplotlib.pyplot as plt
import numpy as np

matriz = np.random.rand(6,6)

plt.imshow(matriz, cmap='hot', interpolation='nearest')
plt.title("Mapa de Calor")
plt.colorbar()
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Ejemplo de matriz de confusión (3 clases)
conf_matrix = np.array([[50, 2, 1],
                        [4, 45, 5],
                        [2, 3, 60]])

fig, ax = plt.subplots()
cax = ax.matshow(conf_matrix, cmap='Blues')

plt.title("Matriz de Confusión")
plt.xlabel("Predicciones")
plt.ylabel("Clases reales")
plt.colorbar(cax)

# Etiquetas en cada celda
for i in range(conf_matrix.shape[0]):
    for j in range(conf_matrix.shape[1]):
        plt.text(j, i, str(conf_matrix[i, j]),
                 ha='center', va='center', color='black')

plt.show()
```
