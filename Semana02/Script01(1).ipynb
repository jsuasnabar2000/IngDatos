# Módulo NumPy
NumPy es una biblioteca de Python de código abierto orientada al cálculo numérico de alto rendimiento. Su principal aporte es el objeto ndarray, un arreglo multidimensional que permite almacenar y manipular grandes volúmenes de datos de manera eficiente.
Además, ofrece funciones avanzadas para álgebra lineal, estadísticas, transformadas de Fourier y generación de números aleatorios, convirtiéndose en la base del ecosistema científico y de análisis de datos en Python. *NumPy fue creado en 2005 por Travis Oliphant*.

## ¿Por qué usar NumPy?

En Python tenemos listas que sirven al propósito de los arrays, pero son lentas de procesar. NumPy tiene como objetivo proporcionar un objeto de matriz que sea hasta 50x más rápido que Listas tradicionales de Python. El objeto array en NumPy se llama ndarray, Proporciona una gran cantidad de funciones de apoyo que hacen trabajar con ndarrayMuy fácil. Las matrices se utilizan con mucha frecuencia en la ciencia de datos, donde la velocidad y los recursos Son muy importantes.

## ¿Por qué es NumPy más rápido que las listas?
NumPy es más rápido que las listas de Python porque está implementado sobre estructuras de bajo nivel en C, lo que permite un acceso directo y eficiente a la memoria. Mientras que las listas de Python son colecciones heterogéneas y dinámicas que almacenan referencias a objetos, los arrays de NumPy son bloques contiguos de memoria con datos homogéneos, lo que facilita operaciones vectorizadas sin necesidad de bucles explícitos. Además, NumPy aprovecha optimizaciones como broadcasting y funciones matemáticas compiladas, reduciendo la sobrecarga del intérprete de Python y logrando un rendimiento cercano al de lenguajes compilados como C o Fortran.
