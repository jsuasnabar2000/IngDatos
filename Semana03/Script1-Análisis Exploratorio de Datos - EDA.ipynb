# Análisis Exploratorio de Datos (EDA)
## 1. Introducción al Análisis Exploratorio de Datos (EDA)

El Análisis Exploratorio de Datos (Exploratory Data Analysis, EDA) constituye la primera etapa de cualquier proyecto de ciencia de datos. Su propósito es comprender la estructura, calidad, comportamiento y relaciones presentes en los datos antes de construir modelos predictivos o realizar inferencias estadísticas. Este enfoque fue impulsado por el estadístico John W. Tukey, quien propuso que el análisis de datos debía convertirse en una disciplina independiente de la inferencia estadística tradicional.

El EDA emplea:

* Estadísticos descriptivos.
* Métodos gráficos.
* Técnicas de resumen de datos.
* Identificación de patrones y anomalías.

Su finalidad es transformar los datos en información útil para la toma de decisiones y para el desarrollo de modelos analíticos robustos.

## 2. Elementos de los Datos Estructurados

Los sistemas de información y la tecnología de internet han generado gran cantidad de volúmenes de datos, pero en su gran mayoría no están estructurados.
Los datos utilizados en ciencia de datos provienen de múltiples fuentes:

* Sensores.
* Transacciones.
* Redes sociales.
* Imágenes.
* Videos.
* Sistemas empresariales.

Sin embargo, la mayoría de los procedimientos estadísticos requieren que los datos se encuentren en una estructura organizada, generalmente en forma tabular.

### 2.1 Tipos de datos
#### a) Datos numéricos (Numeric Data)

Representan cantidades medibles y pueden ser:

**Datos continuos**

Pueden asumir cualquier valor dentro de un intervalo.

Ejemplos:

* Temperatura.
* Tiempo de procesamiento.
* Velocidad del viento.

**Datos discretos**

Admiten únicamente valores enteros.

Ejemplos:

* Número de estudiantes.
* Cantidad de errores.
* Número de ventas.
#### b) Datos categóricos (Categorical Data)

Representan clases o categorías.

Ejemplos:

* Estado civil.
* Carrera profesional.
* Tipo de dispositivo.

Los datos categóricos incluyen:

**Datos binarios**

Solo poseen dos posibles categorías:

Sí / No
Verdadero / Falso
0 / 1

**Datos ordinales**

Poseen un orden natural.

Ejemplos:

Bajo, Medio, Alto.
Calificación de 1 a 5.
### 2.2 Importancia del tipo de dato

La clasificación de una variable determina:

El método estadístico adecuado.
El tipo de visualización.
El algoritmo de aprendizaje automático.
El tratamiento computacional realizado por herramientas como R y Python.
## 3. Datos Rectangulares

La estructura fundamental en ciencia de datos es la denominada datos rectangulares (Rectangular Data).

Se trata de una matriz bidimensional organizada en:

* Filas → registros.
* Columnas → variables o características.

Ejemplos:

* Hojas de cálculo.
* Tablas SQL.
* DataFrames.

### 3.1 Terminología principal
**Data Frame**

Estructura tabular básica utilizada en R y Python.

**Feature (Característica)**

Variable predictora.

Sinónimos:
* Atributo.
* Predictor.
* Variable independiente.

**Outcome**

Variable objetivo que se desea explicar o predecir.

Sinónimos:

* Target.
* Response.
* Variable dependiente.

**Record**

Cada fila de la tabla.

Sinónimos:

* Caso.
* Observación.
* Instancia.
* Muestra.

### 3.2 DataFrames e índices

En sistemas de análisis de datos:

**Python (Pandas)**

Utiliza objetos DataFrame con índices automáticos y jerárquicos.

**R**

Utiliza objetos data.frame con índices basados en el orden de filas.

Los índices mejoran:

* Búsqueda.
* Filtrado.
* Rendimiento computacional.

## 4. Estructuras de Datos No Rectangulares

Aunque los datos tabulares son predominantes, existen otras estructuras.

### 4.1 Series temporales

Representan observaciones registradas a lo largo del tiempo.

Ejemplos:

* Ventas mensuales.
* Temperatura diaria.
* Producción anual.

### 4.2 Datos espaciales

Incluyen información georreferenciada.

Ejemplos:

* Mapas.
* Sistemas GIS.
* Imágenes satelitales.

### 4.3 Grafos o redes

Representan relaciones entre entidades.

Ejemplos:

* Redes sociales.
* Redes de transporte.
* Sistemas de recomendación.

No obstante, el análisis predictivo generalmente requiere transformar estas estructuras a formato rectangular.

## 5. Estimaciones de Localización

Después de organizar los datos, el siguiente paso consiste en identificar un valor representativo o central de la distribución.

Estas medidas se denominan estimaciones de localización.

### 5.1 Media (Mean)

Es la suma de todos los valores dividida por el número de observaciones.

$\bar{x}=\frac{\sum x_i}{n}$

La media utiliza toda la información disponible.

Ventajas:

* Fácil interpretación.
* Base de numerosos modelos estadísticos.

Limitación:

* Sensible a valores extremos.
### 5.2 Media recortada (Trimmed Mean)

Consiste en eliminar un porcentaje de valores extremos antes de calcular la media.

Ventajas:

* Reduce la influencia de outliers.
* Mayor robustez frente a distribuciones asimétricas.
### 5.3 Media ponderada (Weighted Mean)

Asigna pesos específicos a cada observación.

$\bar{x}_w=\frac{\sum w_i x_i}{\sum w_i}$

Aplicaciones:

* Encuestas.
* Sensores con distinta precisión.
* Datos desbalanceados.
### 5.4 Mediana (Median)

Valor ubicado en el centro de una distribución ordenada.

Propiedades:

* Muy resistente a valores extremos.
* Adecuada para distribuciones sesgadas.
### 5.5 Mediana ponderada

Extensión de la mediana que incorpora pesos para cada observación.

### 5.6 Percentiles

Dividen una distribución ordenada en segmentos.

Ejemplos:

* Percentil 25.
* Percentil 50 (mediana).
* Percentil 75.

Permiten comprender mejor la posición relativa de los datos.

### 5.7 Robustez y valores atípicos
**Robustez**

Capacidad de una medida para resistir el efecto de observaciones extremas.

**Valor atípico (Outlier)**

Observación que se encuentra alejada del comportamiento general de los datos.

Los outliers pueden deberse a:

* Errores de medición.
* Registros erróneos.
* Fenómenos reales poco frecuentes.

Su identificación constituye una tarea fundamental del EDA.

## 6. Ejemplo: Estimaciones de Localización para Población y Tasas de Homicidio

El capítulo presenta un conjunto de datos de los estados de Estados Unidos que incluye:

* Población.
* Tasa de homicidios.

Los resultados muestran que:

$Media>Media Recortada>Mediana$

Lo que evidencia la existencia de estados con poblaciones extremadamente grandes que influyen sobre la media.

Asimismo, para variables agregadas como la tasa nacional de homicidios se recomienda emplear:

* Media ponderada.
* Mediana ponderada.

Estas consideran el tamaño poblacional de cada estado.
