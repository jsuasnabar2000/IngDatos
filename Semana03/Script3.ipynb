## 7. Exploración de la Distribución de Datos
Una vez obtenidas las medidas de localización (media, mediana) y variabilidad (desviación estándar, rango intercuartílico), el siguiente paso consiste en comprender cómo se distribuyen los datos en su conjunto. El objetivo es identificar patrones de concentración, dispersión, asimetrías y posibles valores atípicos.

### 7.1 Boxplots y Percentiles
Los percentiles permiten dividir una distribución ordenada en segmentos porcentuales.

**Ejemplo:**
* Percentil 25 (Q1)
* Percentil 50 (Mediana)
* Percentil 75 (Q3)

La diferencia entre Q3 y Q1 constituye el Rango Intercuartílico (IQR), una medida robusta de dispersión.

**Boxplot**
El diagrama de caja y bigotes (Boxplot), introducido por John Tukey, resume visualmente la distribución de los datos.

*Sus componentes son:*
* Línea central: mediana.
* Parte inferior de la caja: percentil 25.
* Parte superior de la caja: percentil 75.
* Bigotes: rango típico de observaciones.
* Puntos aislados: posibles valores atípicos.

*Ventajas*
* Detecta asimetrías.
* Identifica valores extremos.
* Facilita comparaciones entre grupos.

### 7.2 Tablas de Frecuencia e Histogramas
**Tabla de frecuencia**
Es una tabla que contabiliza cuántas observaciones pertenecen a determinados intervalos o categorías.

*Ejemplo:*

| Intervalo | Frecuencia |
|-----------|------------|
| 0-10      | 15         |
| 10-20     | 32         |
| 20-30     | 28         |

**Histograma**
Es la representación gráfica de una tabla de frecuencias.

*Características:*
* Eje X: intervalos de valores.
* Eje Y: frecuencia.

*Barras contiguas.*
Permite observar concentración y dispersión.

*Aplicaciones*
* Identificar sesgos.
* Detectar multimodalidad.
* Analizar simetría de la distribución.

### 7.3 Gráficos de Densidad

Un **Density Plot** es una versión suavizada del histograma.

En lugar de mostrar conteos discretos:
* Estima una función continua de probabilidad.
* Facilita la identificación de tendencias.
* Reduce la dependencia del tamaño de los intervalos.

**Ventajas frente al histograma**
* Menor sensibilidad al ancho de los bins.
* Mejor visualización de patrones complejos.
* Comparación más sencilla entre múltiples distribuciones.

### 7.4 Ideas Clave
La exploración de distribuciones permite:
* Analizar forma y comportamiento de los datos.
* Detectar anomalías.
* Comparar grupos de observaciones.
* Comprender la estructura antes del modelado estadístico.

## 8. Exploración de Datos Binarios y Categóricos
Los datos categóricos representan clases o categorías en lugar de magnitudes numéricas. Su análisis requiere técnicas diferentes a las utilizadas para variables continuas.

### 8.1 Moda
**La moda (Mode)** es la categoría más frecuente dentro de un conjunto de datos.

*Ejemplo:*
| Estado Civil | Frecuencia |
|---|---:|
| Soltero | 45 |
| Casado | 60 |
| Viudo | 10 |

* Moda = Casado

La moda constituye la medida de tendencia central más importante para variables categóricas.

### 8.2 Valor Esperado
Cuando las categorías pueden asociarse con valores numéricos, es posible calcular el valor esperado (Expected Value).

*Se calcula como:*

$EV=\sum (Valor_i \times Probabilidad_i)$

*Aplicaciones*
* Evaluación financiera.
* Gestión de riesgos.
* Marketing.
* Modelos de inversión.

### 8.3 Probabilidad
La probabilidad representa la frecuencia relativa con la que ocurriría un evento si el experimento se repitiera infinitamente.

Ejemplos:

Probabilidad de lluvia.
Probabilidad de fraude.
Probabilidad de conversión de clientes.

La probabilidad constituye la base conceptual de gran parte de la estadística moderna.

### 8.4 Gráficos de Barras
Los bar charts son la principal herramienta de visualización de datos categóricos.

Características:
* Eje X: categorías.
* Eje Y: frecuencias o porcentajes.
* Barras separadas entre sí.

Se utilizan para:
* Comparar categorías.
* Visualizar distribución de frecuencias.
* Detectar categorías dominantes.

### 8.5 Diferencia entre Histograma y Gráfico de Barras
| Histograma | Barras                   |
| ------------|--------------------------|
| Datos numéricos |	Datos categóricos    |
| Barras contiguas | Barras separadas   |
| Intervalos continuos | Categorías discretas |


## 9. Correlación
La correlación mide el nivel de asociación existente entre variables numéricas.

### 9.1 Tipos
Dos variables presentan:

**Correlación positiva**
Cuando aumentan simultáneamente.

Ejemplo:
Horas de estudio y rendimiento académico.

**Correlación negativa**
Cuando una aumenta y la otra disminuye.

Ejemplo:
Velocidad y tiempo de viaje.

**Correlación nula**
Cuando no existe relación observable.

### 9.2 Coeficiente de Correlación de Pearson
Su valor varía entre:

$−1≤r≤1$

Interpretación:

| Valor |	Interpretación                |
|-------|-------------------------------|
| +1	  | Correlación perfecta positiva |
| 0	    | Sin correlación               |
| -1	  | Correlación perfecta negativa |


### 9.3 Matriz de Correlación

Cuando existen múltiples variables, la correlación puede resumirse mediante una matriz de correlación.

La matriz contiene:

* Variables en filas.
* Variables en columnas.
* Coeficientes de correlación en cada intersección.

Aplicaciones
* Selección de variables.
* Reducción de dimensionalidad.
* Detección de redundancias.

### 9.4 Diagramas de Dispersión
El scatterplot constituye la principal visualización de la correlación.

Características:

Cada punto representa una observación.
Eje X = Variable 1.
Eje Y = Variable 2.

Permite identificar:
* Tendencias lineales.
* Agrupamientos.
* Valores extremos.
* Relaciones no lineales.

### 9.5 Limitaciones
Una correlación elevada:

✅ Indica asociación.
❌ No implica causalidad.

Dos variables pueden correlacionarse debido a factores externos no observados.

## 10. Exploración de Dos o Más Variables
El análisis multivariable busca descubrir relaciones simultáneas entre varias variables.

### 10.1 Análisis Bivariado y Multivariado
**Bivariado**
Analiza la relación entre:
Dos variables.

Ejemplos:

Edad y salario.
Temperatura y ventas.

**Multivariado**
Analiza:
Tres o más variables.

Ejemplo:
Edad, educación, ingreso y desempeño laboral.

### 10.2 Hexagonal Binning
Cuando existen cientos de miles de observaciones, un scatterplot tradicional se vuelve ilegible.

La técnica Hexagonal Binning agrupa puntos dentro de hexágonos.

**Ventajas**
Reduce saturación visual.
Mantiene patrones generales.
Adecuada para Big Data.

### 10.3 Contour Plots

Los gráficos de contornos muestran la densidad conjunta de dos variables numéricas.

Funcionan de manera similar a los mapas topográficos:

Curvas cerradas representan concentraciones de datos.
Zonas densas aparecen como “picos”.

**Permiten identificar:**
Agrupamientos.
Modos múltiples.
Subpoblaciones.

### 10.4 Tablas de Contingencia
Las contingency tables resumen la relación entre variables categóricas.

Ejemplo:

Estado Civil	Compra	No compraSoltero	120	80
Casado	150	50

Permiten analizar:

Dependencias.
Distribuciones conjuntas.
Frecuencias relativas.

### 10.5 Variables Numéricas y Categóricas

Cuando se combina:
* Una variable categórica.
* Una variable numérica.

Se utilizan principalmente:

*Boxplots*
Permiten comparar distribuciones entre categorías.

*Ejemplo:*
* Salario por profesión.
* Rendimiento por método educativo.
* Violin Plots

Extienden al boxplot mostrando también la densidad de la distribución.

*Ventajas:*
* Mayor detalle.
* Visualización de multimodalidad.
* Comparación más rica de grupos.


### 10.6 Visualización de Múltiples Variables
La visualización multivariable utiliza:
* Paneles condicionados (Facets).
* Hexagonal Binning.
* Matrices de correlación.
* Violin Plots.
* Gráficos de dispersión segmentados.

Estas técnicas permiten estudiar simultáneamente:
* Variables predictoras.
* Variables categóricas.
* Relaciones complejas entre factores.
